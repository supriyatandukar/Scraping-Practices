```python
from selenium import webdriver
from bs4 import BeautifulSoup
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# s = Service('chromedriver-win64/chromedriver.exe')
# driver = webdriver.Chrome(service=s)
driver = webdriver.Chrome()

try:
    dynamic_url = "http://quotes.toscrape.com/js/"; #js nai use wala
    driver.get(dynamic_url)

    wait = WebDriverWait(driver, 10)
    wait.until(EC.presence_of_element_located((By.CLASS_NAME, "container"))) #webdriver kai code ho

    print("Page loaded and quotes are present.")

    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'lxml')
   
    quotes = soup.find_all('div', class_='quote')
    print(f"Found {len(quotes)} quotes using Selenium.")
   
    for quote in quotes:
       
        text = quote.find('span', class_='text').text
        author = quote.find('small', class_='author').text
        print(f"- '{text}' by {author}")
       
finally:
    # Always close the browser window!
    if 'driver' in locals():
        driver.quit()
        print("\nBrowser closed.")
```

```python
#selenium le next next page lagxa photograph jastaii
import time
from selenium import webdriver
from selenium.common.exceptions import WebDriverException, TimeoutException
from bs4 import BeautifulSoup
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from urllib.parse import urljoin
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

chrome_options = Options()
#chrome_options.add_argument('--headless')  #open hudaina background mai sab kaam hunxa
chrome_options.add_argument('--disable-gpu')
chrome_options.add_argument('--no-sandbox')
chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36")
chrome_options.add_argument("--start-maximized")

driver = None
try:
    driver = webdriver.Chrome(options=chrome_options)
    base_url = "https://news.ycombinator.com/"
    all_news_data = []
    page_count = 0
    max_pages = 5
    current_page_url = "https://news.ycombinator.com/"
    retry_count = 0
    max_retries = 3

    while current_page_url and page_count < max_pages:
        try:
            page_count += 1
            print(f"Scraping page {page_count}...")
            
            # Check if driver is still connected
            try:
                driver.current_url
            except WebDriverException:
                print("Driver disconnected, creating new one...")
                driver = webdriver.Chrome(options=chrome_options)
           
            driver.get(current_page_url)
            
            # Increased wait time
            wait = WebDriverWait(driver, 15)
            wait.until(EC.presence_of_element_located((By.CLASS_NAME, "athing")))

            page_source = driver.page_source
            soup = BeautifulSoup(page_source, 'lxml')
           
            news = driver.find_elements(By.CLASS_NAME, "athing")

            for q in news:
                try:
                    text = q.find_element(By.CLASS_NAME, "titleline").text
                    all_news_data.append(text)
                except:
                    continue
           
            more_button = soup.find('a', class_='morelink')
            if more_button:
                next_page_relative_url = more_button['href']
                current_page_url = urljoin(base_url, next_page_relative_url)
                time.sleep(5)  # Increased sleep to be respectful to the server
            else:
                current_page_url = None
                
            retry_count = 0  # Reset retry on successful page

        except (TimeoutException, WebDriverException) as e:
            retry_count += 1
            print(f"Error on page {page_count}: {str(e)[:100]}")
            
            if retry_count < max_retries:
                print(f"Retrying... (Attempt {retry_count}/{max_retries})")
                time.sleep(10)  # Wait longer before retry
            else:
                print(f"Max retries reached for page {page_count}. Moving to next...")
                page_count += 1
                retry_count = 0

    print("\n" + "="*50)
    for i, news_item in enumerate(all_news_data, 1):
        print(f"{i}. {news_item}")
    print(f"\nTotal news items scraped: {len(all_news_data)}")
    
except Exception as e:
    print(f"Fatal error occurred: {e}")
    
finally:
    if driver:
        try:
            driver.quit()
            print("\nBrowser closed.")
        except:
            print("\nBrowser already closed.")
```

```python
import pandas as pd

# Create DataFrame from scraped data
df = pd.DataFrame({
    "News": all_news_data
})
```

```python
# Print the dataframe
print(df)
print(f"\nShape: {df.shape}")
```

```python

# Save to CSV
output_filename = 'scraped_news.csv'
df.to_csv(output_filename, index=False)
print(f"\nData successfully saved to {output_filename}!")
```
