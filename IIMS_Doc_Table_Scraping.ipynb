```python
import requests
import time
from bs4 import BeautifulSoup
import pandas as pd
print("Libraries imported successfully!")
```

```python
URL = "https://iimscollege.edu.np/program/bachelor-of-computer-science-hons/"
headers = {
    "User-Agent": "My College Web Scraper"
}

response = requests.get(URL, headers=headers)
```

```python
print(response)
```

```python
print(response.content)
```

```python
soup = BeautifulSoup(response.text, 'lxml')
```

```python
table = soup.find("h2", string = "Documents Required")
print(f"Table Heading: {table.text}")
```

```python
doc_table = table.find_next("table")
print(doc_table.text)
```

```python
print(doc_table.prettify())
```

```python
all_th = doc_table.find_all("th")
all_th
```

```python
table_heading = [th.text.strip() for th in doc_table.find_all("th")]
print(table_heading)
```

```python
rows = []
for tr in doc_table.find_all("tr"):
    cols = [td.text.strip() for td in tr.find_all("td")]
    if cols:
        rows.append(cols)
```

```python
df = pd.DataFrame(rows, columns=table_heading)
```

```python
df
```

```python
output_filename = 'iims_doctable.csv'
df.to_csv(output_filename, index=False)
print(f"\nData successfully saved to {output_filename}!")
```
