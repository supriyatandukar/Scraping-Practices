```python
import requests
import time
from bs4 import BeautifulSoup
import pandas as pd
print("Libraries imported successfully!")
```

```python
URL = "https://iimscollege.edu.np/program/bachelor-of-computer-science-hons/"
headers = {
    "User-Agent": "IIMS College Web Scrapper"
}

# Make the request
response = requests.get(URL, headers=headers)
```

```python
print(response)
```

```python
#Check the status code
if response.status_code == 200:
    print("Success! Request was successful.")
    print("\nFirst 500 characters of the page:")
    print(response.text[:500])
else:
    print(f"Failed to retrieve the page. Status code: {response.status_code}")
```

```python
#Create a Beautiful Soup object
soup = BeautifulSoup(response.text, 'lxml')
```

```python
print(soup)
```

```python
print(soup.prettify())
```

```python
image = soup.find_all('img')

for img in image:
    img_url=img.get('src')
    if img_url:
        print(img_url)
```

```python
course_img = soup.find_all('div', class_='specification-course-card')
```

```python
print(f"Cards found after browser rendering: {len(course_img)}")
```

```python
for i in course_img:
    img_tag = i.find('img')
    if img_tag and img_tag.get('href'):
        print(img_tag['href'])
```

```python
print("specification-course-card" in response.text)
```

```python
course_img
```

```python
course_data_img=[]
for i in course_img:
    img_tag = i.find('img')

    if img_tag:
        img_url = img_tag.get('data-src')
        print(img_url)

        if img_url:
            course_data_img.append(img_url)
```

```python
course_data_img
```

```python
df = pd.DataFrame({'Image_URL': course_data_img})
```

```python
output_filename = 'iims_doctable_Courses.csv'
df.to_csv(output_filename, index=False)
print(f"\nData successfully saved to {output_filename}!")
```
